````python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Read Inside Airbnb data") \
    .getOrCreate()
````

The code **reads a compressed CSV file into a Spark DataFrame using specific configuration options to handle data formatting and parse errors**.

## Spark CSV Options

* **header:** Tells Spark whether the first row of the file contains column names (`True` uses the first row as headers).
* **inferschema:** Tells Spark to automatically guess the data type (like integer or string) for each column instead of treating everything as text.
* **sep:** Defines the character used to separate individual values or columns in the file (such as a comma).
* **quote:** Sets the character used to enclose values, allowing data fields to safely contain separator characters or spaces or new lines(multiline fields).
* **escape:** Sets the character used to unblock or escape special characters inside a data field.
* **multiline:** Allows a single row of data to span across multiple lines if it contains a line break inside a quoted field.
* **mode:** Tells Spark how to handle bad or corrupted records (`PERMISSIVE` keeps bad rows by setting malformed fields to null instead of crashing).

````python
listings = spark.read.csv("../data/input/listings.csv.gz",
                          header = True,
                          inferSchema=True,
                          sep=",",
                          quote='"',
                          escape='"',
                          multiLine=True,
                          mode="PRMISSIVE",
                         )
````

## mulitline field in csv
put all the chars inside `"<cell-data>"` while writing multiline field

To use literal double quotes inside a CSV field that is already surrounded by quotes, you must escape them by typing two double quotes ("") in a row.

When a CSV reader parses the file, it will interpret those two consecutive quotes as a single, literal double quote character inside that cell.

### Code Example

Here is how you format text that contains both quotes and line breaks in raw CSV text:

```csv
ID,Topic,Quote_Text
1,Inspiration,"The speaker said, ""This is line one of the quote.
And this is line two of the same quote."""
2,Business,"Always follow the ""gold standard"" rules."
```

### How It Looks in a Spreadsheet

When you import the raw text above into **Microsoft Excel** or **Google Sheets**, the software automatically removes the outer wrapper and collapses the double-quotes into single ones:

| **ID** | **Topic** | **Quote_Text** |
|---|---|---|
| 1 | Inspiration | The speaker said, "This is line one of the quote. And this is line two of the same quote." |
| 2 | Business | Always follow the "gold standard" rules. |

### Quick Summary of Rules

* **Wrap the entire cell** in double quotes if it contains commas, line breaks, or internal quotes.
* **Double up internal quotes** (`""`) whenever you want a literal `"` to show up in your final data.
